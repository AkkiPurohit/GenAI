```python
# @tool decorator
from langchain_core.tools import tool

@tool
def division(a:int,b:int)->int:
    """Divide by 2 number"""
    return a/b

```

```python
print(division.name)
print(division.description)
print(division.args)
```

```python
## async implementation
from langchain_core.tools import tool

@tool
async def division(a:int,b:int)->int:
    """Divide by 2 number"""
    return a/b
```

```python
from typing import Annotated,List

@tool
def  multiply_by_max(
    a: Annotated[int,"A value"],
    b: Annotated[List[int],"List of ints over which to take maximum"]
)->int:
    """Multiply a by the max of b"""
    return a*max(b)
```

```python
print(multiply_by_max.name)
print(multiply_by_max.description)
print(multiply_by_max.args)
```

### Structured Tool

The StructuredTool.from_function class methods provides a bit more configurability than the @tool decorator,without requiring much addition code.

```python
from langchain_core.tools import StructuredTool
def multiply(a:int,b:int)->int:
    """multiply by 2 number"""
    return a*b


async def amultiply(a:int,b:int)->int:
    """multiply by 2 number"""
    return a*b

calculator=StructuredTool.from_function(func=multiply,coroutine=amultiply)

print(calculator.invoke({"a":2,"b":3}))
print(await calculator.ainvoke({"a":2,"b":5}))
```

### Call Tool With LLM:

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

api_wrapper=WikipediaAPIWrapper(top_k_results=5,doc_content_chars_max=500)
wiki_tool=WikipediaQueryRun(api_wrapper=api_wrapper)

print(wiki_tool.invoke({"query":"langchain"}))
```

```python
from langchain_core.tools import tool

@tool
def add(a:int,b:int)->int:
    "Adds a and b"
    return a+b

@tool
def multiply(a:int,b:int)->int:
    "Multiply a and b "
    return a*b
```

```python
from langchain_tavily import TavilySearch
tavily_tool = TavilySearch(

max_results=5,

topic="general",

# include_answer=False,

# include_raw_content=False,

# include_images=False,

# include_image_descriptions=False,

# search_depth="basic",

# time_range="day",

# include_domains=None,

# exclude_domains=None
)
```

```python
tools=[wiki_tool,add,multiply,tavily_tool]
```

```python
tools
```

```python
from langchain.chat_models import init_chat_model
llm=init_chat_model("openai/gpt-oss-120b",model_provider="groq")
llm
```

```python
llm_with_tools=llm.bind_tools(tools,parallel_tool_calls=True)
llm_with_tools
```

```python
from langchain_core.messages import HumanMessage
query="what is today sports news"

messages=[HumanMessage(query)]
response=llm_with_tools.invoke(query)
print(response)
```

```python
response.tool_calls
```

```python
for tool_call in response.tool_calls:
    selected_tool={"add":add,"multiply":multiply,"wikipedia":wiki_tool,"tavily_search":tavily_tool}[tool_call["name"].lower()]
    tool_msg=selected_tool.invoke(tool_call)
    messages.append(tool_msg)
messages
```

```python
from langchain_core.messages import HumanMessage,SystemMessage
query="Who was Alan Turing and what is the score of india vs newzeland test match today"
msg=[
    SystemMessage("Never answer from memory. Use the wikipedia tool for facts about "
                  "people, places or history, and tavily_search for current events. "
                  "Call all the tools you need in one step."),
    HumanMessage(query)
    ]

tools_by_name={"add":add,"multiply":multiply,"wikipedia":wiki_tool,"tavily_search":tavily_tool}
called_tool=[]
while True:
    res = llm_with_tools.invoke(msg)     # ask the model
    msg.append(res)                      # save its reply

    if not res.tool_calls:               # no tool requested -> done
        break

    called_tool.append(res.tool_calls)
    for tc in res.tool_calls:            # run each requested tool
        tool = tools_by_name[tc["name"].lower()]
        msg.append(tool.invoke(tc))      # save the tool result

print(res.content)
```
